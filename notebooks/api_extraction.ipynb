```python
import requests
import pandas as pd
import os
import json
from dotenv import load_dotenv
import time
```

```python
dotenv_path = os.path.join(os.path.abspath('..'), '.env')
load_dotenv(dotenv_path=dotenv_path)
```

```python
api_key = os.getenv("API_Key")
api_URL = "https://serpapi.com/search"
```

```python
KEYWORDS = [
    "smartphone India",
    "laptop India",
    "washing machine India",
    "air conditioner AC India",
    "office chair India",
    "shoes India",
    "smartwatch India",
    "iron box India",
    "bluetooth speaker India",
    "headphones headset India",
    "computer mouse India",
    "tablet India",
    "monitor India",
    "keyboard India",
    "camera India",
    "smartwatch watch India",
    "power bank India",
    "wireless earbuds India",
    "gaming mouse India",
    "usb cable India",
]
print(f"Total keywords  : {len(KEYWORDS)}")
print(f"Products/page   : 40")
print(f"Pages/keyword   : 3")
print(f"Expected total  : {len(KEYWORDS)} × 40 × 3 = {len(KEYWORDS) * 40 * 3} products")
```

```python
def fetch_prod_per_page(keyword,page_number,api_key,num_per_page):
    start = page_number * num_per_page   # calculate offset
    
    params = {
        'q':        keyword,
        'engine':   'google_shopping',
        'api_key':  api_key,
        'num':      num_per_page,
        'start':    start,          # ← this is the pagination parameter
        'hl':       'en',
        'gl':       'in',
        'location': 'India',
        }
    
    api_response = requests.get(url=api_URL,params=params)
    
    if api_response.status_code == 200:
        print("Response received successfuly")
        return api_response.json()
    else:
        print(f"Error{api_response.status_code} for the keyword {keyword} in the page {page_number+1}")
        return {}
```

```python
def extract_feilds_from_response(keyword,api_response):
    products = []
    
    items = api_response.get('shopping_results', [])
    
    for item in items:
            product = {
                'keyword':   keyword,
                'title':     item.get('title',            ''),
                'price':     item.get('extracted_price',  None),  # number format
                'raw_price': item.get('price',            ''),    # text format e.g. "₹79,999"
                'rating':    item.get('rating',           None),
                'reviews':   item.get('reviews',          None),
                'platform':  item.get('source',           ''),    # e.g. "Amazon.in"
                'position':  item.get('position',         None),  # rank in results
                'delivery':  item.get('delivery',         ''),
                'link':      item.get('link',             ''),
                'thumbnail': item.get('thumbnail',        ''),
            }
            products.append(product)
    
    return products
        
        
```

```python
def extract_all_products(keywords,pages = 3,num_per_page=40):
    all_products = []
    
    total = len(keywords)
    for i, keyword in enumerate(keywords,1):
        print(f"{1/total},keyword {keyword}")

        for page in range(pages):
            response_data = fetch_prod_per_page( keyword,page, api_key, num_per_page)
            products = extract_feilds_from_response(keyword,response_data)
            all_products.extend(products)
            print(f"  Page {page+1}: {len(products)} products")
            time.sleep(0.5)   # wait 0.5 seconds between requests
            
    print(f"\nDone! Total products collected: {len(all_products)}")
    return all_products
    
products = extract_all_products(KEYWORDS,pages=3,num_per_page=40)
df = pd.DataFrame(products)

df.to_csv(r"C:\Users\ADMIN\Documents\subem\2026\Guvi\Project\brand\data\raw\pandas_shopping_results.csv")
            
```
